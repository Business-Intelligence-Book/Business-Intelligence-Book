<p style="margin:0 0 6px 0"><b style="color:#2F5597">⭐ Fun fact: More degrees can make things worse</b></p>
In 1901 the German mathematician Carl Runge showed that a polynomial of high degree can pass exactly through a set of equally spaced points and still be far from the function behind them: it oscillates wildly near the ends of the interval. This is known as Runge's phenomenon, and it is one reason why the test error grows when the polynomial degree is too high.
