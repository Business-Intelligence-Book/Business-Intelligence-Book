### Column types

Each column of a dataset can have a different type. The table below lists the data types of pandas and how they relate to Python and NumPy types.

| pandas dtype | Python type | NumPy type | Usage |
| :--- | :--- | :--- | :--- |
| object | str or mixed | string_, unicode_, mixed types | Text, or mixed numeric and non-numeric values |
| int64 | int | int_, int8, int16, int32, int64, uint8, uint16, uint32, uint64 | Integer numbers, e.g. 1, 2, 3 |
| float64 | float | float_, float16, float32, float64 | Floating point numbers (with decimals) |
| bool | bool | bool_ | True/False values |
| datetime64 | NA | datetime64[ns] | Date and time values |
| timedelta[ns] | NA | NA | Differences between two datetimes |
| category | NA | NA | Finite list of text values |

pandas infers the correct type most of the time, but it is important to check the columns and convert them if needed. Otherwise the results can be wrong.

The `.dtypes` attribute shows the type of each column:

````python
data.dtypes
````

Most columns have the correct type, but we can still improve the DataFrame. Let's convert the `my_datetime` column to the `datetime64` type, so that pandas knows that it contains dates and times.

The [`to_datetime()`](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.to_datetime.html) function does this.

````python
# First, let's convert the column type
data["my_datetime"] = pd.to_datetime(data["my_datetime"])

# Now let's have a look at the dtypes once again
data.dtypes
````

The `user_id` column seems to contain integers. pandas did not infer this, because some rows are corrupted (they contain values that are not integers).

We can still convert the column with `pd.to_numeric`. The argument `errors='coerce'` sets all corrupted entries to NaN (Not a Number).

The other possible values for `errors` are:
 * `errors='raise'`: raise an error at the first invalid entry.
 * `errors='ignore'`: return the original data without any conversion if there is an invalid entry.

````python
# Let's convert the column type
data["user_id"] = pd.to_numeric(data["user_id"], errors="coerce")

# Now let's have a look at the dtypes once again
data.dtypes
````

Since the number of possible countries is limited, we can convert that column to the categorical type with the `.astype()` function:

````python
# Change dtype
data["country"] = data["country"].astype("category")

# Now let's have a look at the dtypes once again
data.dtypes
````

### Dropping missing values

The corrupted cells are now NaN.
The function `.isnull()` returns a DataFrame that contains:
 * `False` where the data is present
 * `True` where the data is missing

We can then use `.sum()` to count the missing values in each column.

````python
# Let's see how many column values are missing
data.isnull().sum()
````

We can now remove the missing values with `.dropna()`.
Without arguments, this function drops every row in which at least one value is missing.

````python
# We first drop all the rows where at least one of the values is missing
data = data.dropna()

# Let's have another look
data.isnull().sum()
````

## 2. Loading JSON files

A lot of the data on the Internet comes in JSON format, a structured text format that looks very much like a Python dictionary.

Let's see how to load a JSON dataset into a pandas DataFrame.

We use a snapshot from the Citi Bike API, which gives the status of the bike-sharing stations in New York.

````python
# Import libraries
import pandas as pd

# Display all columns
pd.set_option("display.max_columns", None)
# Display max 50 rows
pd.set_option("display.max_rows", 50)
````

````python
# Read json file

# Import urllib library
from urllib.request import urlopen

# Import json
import json

# Store the URL in url as
# parameter for urlopen
url = "https://raw.githubusercontent.com/Business-Intelligence-Book/business-intelligence-book.github.io/main/labs/data/station_status/station_status.json"

# Store the response of URL
response = urlopen(url)

# Store the JSON response
# from url in data
data = json.loads(response.read())

# Uncomment to see how the data looks:
# print(data)
````

````python
# Below you see how the JSON file looks. The JSON results contain three keys:
#  - The 'data' key contains all the data
#  - The 'last_updated' key contains the timestamp of the last update
#  - And the 'ttl' key contains the time to live of the data
data.keys()
````

````python
# The field "data" contains the information we want but it has a nested structure
data["data"].keys()
# The field "stations" contains the actual data
````

````python
# We can convert the nested dictionary to a pandas dataframe
# We want to extract the data from the key "data"
df_info = pd.DataFrame(data["data"]["stations"])
df_info.sample(5)
````

````python
# Display types of each column
df_info.dtypes
````

````python
# Replace NAN value of all integer fields by 0
df_info.fillna(0, inplace=True)
````

## Grouping values by feature

To compute an indicator (mean, max, min, ...) separately for each value of a feature, you first group the rows by that feature with the pandas function `groupby()`. For example, to compare the average number of available bikes at renting stations (`is_renting = 1`) and at the other stations, we group the rows by `is_renting`.

````python
# Take the mean of bikes available for renting stations (is_renting = 1) and non-renting stations (is_renting = 0)
df_info_mean = df_info.groupby("is_renting")["num_bikes_available"].mean()
df_info_mean.head()
````

__Question 2.1:__ Add a column named `ebike_prop` that contains the proportion of available e-bikes among all available bikes, that is: $$\frac{num \ ebikes \ available}{num \ ebikes \ available + num \ bikes \ available}$$

````python
# Add column with the proportion of e-bikes available
df_info["ebike_prop"] = (
    df_info["num_ebikes_available"]
    / (df_info["num_ebikes_available"] + df_info["num_bikes_available"])
    * 100
)
df_info.head()
````

__Question 2.2:__ Which stations have the greatest proportion of available e-bikes? Display the first 5 stations with their `station_id`.

````python
df_info.sort_values(by="ebike_prop", ascending=False)[
    [
        "station_id",
        "ebike_prop",
        "num_docks_available",
        "num_ebikes_available",
        "num_bikes_available",
    ]
].head(5)
````

__Question 2.3:__ Which stations have the greatest proportion of available e-bikes among those with at least 10 available e-bikes (to maximize your chances of finding an e-bike when you reach the station)? Display the first 10 stations with their id, the proportion, the number of available docks and the number of available bikes.

````python
df_temp = df_info.sort_values(by="ebike_prop", ascending=False)[
    [
        "station_id",
        "ebike_prop",
        "num_docks_available",
        "num_ebikes_available",
        "num_bikes_available",
    ]
]
df_temp[df_temp["num_ebikes_available"] >= 10].head(10)
````

__Question 2.4:__ Simple statistics about a column, such as the mean or the highest values, are often useful. What proportion of the stations have at least 50% of e-bikes among their available bikes? And at least 75%?

````python
# Display the statistics of the initial data file for the newly_created column "ebike_prop"
df_info["ebike_prop"].describe()
````
