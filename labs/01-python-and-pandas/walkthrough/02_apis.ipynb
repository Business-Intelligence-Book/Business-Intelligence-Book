# Introduction

>This walkthrough shows how to get data from the Internet with API requests. You will see how to request data in `json` format from a website, and then how to retrieve it in Colab with a few lines of Python, so that you can work on it.

> We use **2 examples**: one retrieves IMDb data about your **favorite movie**, the other retrieves the **current weather** in any city in the world.

# 1. Data from your favorite movie

In this example, you get the data about a movie of your choice in two steps:

1. Send a request to an API to obtain the data.
2. Load this data in Colab.

## Making an OMDb API request

**Let us take a look at the [OMDb API](https://www.omdbapi.com/).**

> You make an HTTP request with this URL: ```http://www.omdbapi.com/?apikey=[yourkey]&```

You need a **key**, which the API uses to control your requests. With OMDb, free requests are limited to 1,000 per day.

> You can get your own key [here](https://www.omdbapi.com/apikey.aspx) by entering your email address.

> You can also use this key: `6be81d95`.

You then add **filters** to the request to tell the API what you are looking for. They are appended to the end of the URL.

For example, to search for the movie "The Big Short" with its full plot:
> The filters are `t=The+Big+Short&plot=full`. Here `t` is the title of the movie (mandatory for this API) and `plot` has the option `full`.

> The filters are always separated by a `&` sign.

The URL of the request is therefore:

```https://www.omdbapi.com/?apikey=6be81d95&t=The+Big+Short&plot=full```

**Try entering that URL in your browser and see what happens!**

### Result

If everything goes well, you get a web page in `json` format.

This is an example of **semi-structured** data.

>However, all of this is still happening in your web browser, where you cannot process the data with a program.

## Loading the data in Colab

You can read more about this process [here](https://colab.research.google.com/github/nestauk/im-tutorials/blob/3-ysi-tutorial/notebooks/APIs/API_tutorial.ipynb), in the tutorial that inspired this walkthrough.

````python
import requests  # Import the requests library

# Query URL
url = ('https://www.omdbapi.com/?apikey=6be81d95&t=The+Big+Short&plot=full')  # Movie 'The Big Short' with a full plot

print(url)

response = requests.get(url)  # Make a GET request to the URL.

# Print status code (and associated text).
print(f"Request returned {response.status_code} : '{response.reason}'")

# Note: status 200 means the request was successful. Search online for documentation about request statuses.

# Print data returned (parsing as JSON)
payload = response.json()  # Parse `response.text` into JSON

import pprint # This library makes the result look good.

pp = pprint.PrettyPrinter(indent=1)
pp.pprint(payload)
````

**You did it!** You can now process the data about the movie.

````python
# Show the actors that played in the movie

payload['Actors']
````

````python
# Show the plot of the movie

plot = payload['Plot']

print(plot)
````

````python
# Count the number of words in the plot
print(type(plot))

plot_length = plot.split() # Use the split method on a string

print(len(plot_length)) # Count the length of the list containing all words
````

## Current weather in Lausanne

We have already generated an API key: `APPID=c3b012a09cda741d210d8739c796c3da`

The only parameters here are the name of the city and the country code: `Lausanne,ch`.

You could also give the latitude and longitude, as explained in [the documentation](https://openweathermap.org/current).

> The OpenWeather API returns the following `JSON` data.

````python
url = ('https://api.openweathermap.org/data/2.5/weather?q=Lausanne,ch&APPID=c3b012a09cda741d210d8739c796c3da') # City of Lausanne, CH
````

The next cells ask you for the city and the country code:

````python
url_city = input("Enter the city: ")
````

````python
url_city = url_city.lower().capitalize()

url_country = input("Enter the country code:")
````

````python
url_country = url_country.lower()
url_data = url_city + ',' + url_country

url = ('https://api.openweathermap.org/data/2.5/weather?q={}&APPID=c3b012a09cda741d210d8739c796c3da').format(url_data)
````

We now repeat the same operation and fetch the data from the API, with the city and country you just entered.

````python
import requests  # Import the requests library


print(url)

response = requests.get(url)  # Make a GET request to the URL.

# Print status code (and associated text).
print(f"Request returned {response.status_code} : '{response.reason}'")

# Note: status 200 means the request was successful. Search online for documentation about request statuses.

# Print data returned (parsing as JSON)
payload = response.json()  # Parse `response.text` into JSON

import pprint # This library makes the result look good.

pp = pprint.PrettyPrinter(indent=1)
pp.pprint(payload)
````

_Note_: you can play with this request by changing the city in the URL.

````python
# What is the temperature?
print(type(payload['main']))

main_info = payload['main']
main_info
````

````python
# The previous cell is a dictionary. You can directly obtain the temperature:
payload['main']['temp']
````

````python
# You can also translate this data into a pandas DataFrame (cf. pandas walkthrough)
import pandas as pd

name = payload['name']
main_index = []
main_index.append(name)

df_main = pd.DataFrame(main_info, index=main_index)
df_main
````

````python
# Temperature is in Kelvin- let us convert into Celsius

df_main[['temp', 'feels_like', 'temp_min', 'temp_max']] = df_main[['temp', 'feels_like', 'temp_min', 'temp_max']] - 273.15
df_main
````
