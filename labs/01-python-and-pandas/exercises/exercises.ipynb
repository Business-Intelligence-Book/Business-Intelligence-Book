#### Key parameters of `.dropna()`

- **axis**:
  - `0` (default) → drops rows with NaN values.
  - `1` → drops columns with NaN values.

- **how**:
  - `"any"` (default) → drops a row if **any** of its values is NaN.
  - `"all"` → drops a row only if **all** of its values are NaN.

- **inplace**:
  - `False` (default) → returns a new DataFrame without the NaN rows.
  - `True` → modifies the original DataFrame.
