# Python refresher

## 1. Types and variables


The basic data types are the building blocks of all the more complex data structures in Python:

* Strings: sequences of characters
* Numeric types (integer and float): numbers for computations
* Booleans: `True` and `False`, for logical conditions

A variable gives a name to a value, so that we can use the name instead of the value itself.
Run the code below:

```python
print(type(1))
print(type(0.0))
print(type("Hello"))
print(type(False))
print(type(True))
```

```python
a = "Hello world!"
print("Value of a:", a)
print("The type of a:", type(a))
print("The length of a:", len(a))

# Modify the value
b = "Goodbye"
a = b
print(a)
```

**Note**: the variable `a` contains the string "Hello world!". The function `print()` displays the value of a variable. You can change the value of a variable at any time.

```python
males = 7
females = 10
fraction = males / (males + females)
print("Percentage of men:", 100 * fraction)
print("Percentage of men:", round(100 * fraction, 3))
```

**Note:** the function `round()` rounds a value to a given number of decimals.

## 2. Conditions

The Boolean values `True` and `False` are the logical values. Conditional statements use `if`, `elif` and `else`. If the condition after `if` is true, the code below it is executed. Otherwise Python goes on to test the next conditions.

Comparison operators produce Boolean values:

- `a == b`: a equals b
- `a < b`: a is less than b
- `a <= b`: a is less than or equal to b
- `a != b`: a is not equal to b
- `a > b`: a is greater than b
- `a >= b`: a is greater than or equal to b

Boolean values can be combined with the operators `and`, `or` and `not`.

**Example:**

```python
x = True
y = False
z = False

if y or (x and z):
    print("output1")
if not x or (y and z):
    print("output2")
if (x or z) or (y and z):
    print("output3")
if y != z:
    print("output4")
```

```python
name = "Alice"
age = 9
brothers = 2
sisters = 0
print(name == "Alice" or age == 9)
print(name == "Alice" and not age == 9)
print(name == "Miro" or not age > 5)
print("Not, And, Or")


if age > 5 and len(name) < 10:
    print("The condition is true, so this message will be print!")

if brothers > 1 or sisters > 1:
    print("At least one brother and one sister.")
elif brothers > 1 and sisters == 0:
    print("You have more than one brother and no sisters.")
elif brothers == 0 and sisters > 1:
    print("You have more than one sister and no brothers.")
else:
    print("")
```

## 3. Lists

A list is an ordered sequence of values. It can hold values of any type. Let's see how to define one:

```python
primes = [2, 3, 5, 7]
companies = ["Google", "Apple", "Facebook", "Amazon", "Space X", "Huawei", "Samsung"]
# A list of lists
hands = [["J", "Q", "K"], ["2", "2", "2"], ["6", "A", "K"]]
my_favourite_things = [32, "raindrops on roses", help]
```

You access the elements of a list with square brackets.
Python uses zero-based indexing, so the first element has index 0.

**Example:**

```python
print(companies[0])
print(companies[-1])
print(hands[0][0])
print(primes[len(primes) - 1])
```

*Slicing* gets several values at once. `companies[0:3]` asks for the elements of `companies` from index 0 up to, but not including, index 3.

```python
print("Original list: ", companies)
print(companies[0:3])
print(companies[:3])
print(companies[3:])
print(companies[1:-1])
print(companies[-2:])
```

A list can be modified in several ways. You can assign new values to one or more elements, with indexing and slicing.

You can also use list methods such as:

- `list.pop`: removes and returns the last element of the list
- `list.append`: adds an item to the end of the list

**Example:**

```python
# Change one element
companies[4] = "Monsanto"
print(companies)

# Change 3 elements
companies[:3] = ["G", "A", "F"]
print(companies)

companies.append("Bayer")
print(companies)
companies.pop()
print(companies)
```

## 4. Functions

We have already seen some functions, like `print()`.
A function in code is similar to a mathematical function: it takes arguments (input data), does some computations and can return values.

Built-in functions are great, but soon we need to define our own.

**Example:**

```python
def least_difference(a, b, c):
    """Return the smallest difference between any two numbers
    among a, b and c.
    """
    diff1 = abs(a - b)
    diff2 = abs(b - c)
    diff3 = abs(a - c)
    return min(diff1, diff2, diff3)


least_diff = least_difference(2, 5, 8)
print(" The least difference between (2, 5, 8) is : ", least_diff)

# Compact way
print(" The least difference between (2, 5, 8) is : ", least_difference(2, 5, 8))
```

**Note:** this function computes the smallest difference between three values. It uses the Python function `abs()`, which computes the absolute value, and saves the results in variables.
`return` is a keyword used only in functions. When Python reaches a `return` statement, it leaves the function immediately and passes the value on the right-hand side back to the caller.

We can call the function with specific values and save the result in a variable, to use it later in the code.

```python
help(least_difference)
```

Some functions do not return a value. Calling them gives the special value `None`.

**Example**:

```python
def least_difference(a, b, c):
    """Compute the smallest difference between any two numbers
    among a, b and c.
    """
    diff1 = abs(a - b)
    diff2 = abs(b - c)
    diff3 = abs(a - c)
    min(diff1, diff2, diff3)


print(" The least difference between (2, 5, 8) is : ", least_difference(2, 5, 8))
```

Functions can also be passed as arguments to other functions. This is powerful, even if it feels abstract at first.

**Example**:

```python
def mult_by_five(x):
    return 5 * x


def call(fn, arg):
    """Call fn on arg"""
    return fn(arg)


def squared_call(fn, arg):
    """Call fn on the result of calling fn on arg"""
    return fn(fn(arg))


print(
    call(mult_by_five, 1),
    squared_call(mult_by_five, 1),
    sep="\n",  # '\n' is the newline character - it starts a new line
)
```

## 5. Loops

A loop executes some code repeatedly. The `for` statement gives the name of a variable and the set of values to loop over, linked by the word `in`.
The object to the right of `in` can be any object that supports iteration: if it can be thought of as a group of things, you can probably loop over it.

**Example:**

```python
companies = ["Google", "Apple", "Facebook", "Amazon", "Space X", "Huawei", "Samsung"]
for company in companies:
    print(company, end=" - ")
```

`range()` is a function that returns a sequence of numbers. It is very useful for writing loops.

```python
for i in range(5):
    print("i =", i)
```

The other type of loop in Python is the `while` loop, which repeats as long as a condition is true:

```python
i = 0
while i < 10:
    print(i, end=" ")
    i += 1
```

The code below, of the form `[n**2 for n in range(10)]`, is called a list comprehension. It is a very useful feature of Python that makes code shorter.

```python
# List comprehension
squares = [n**2 for n in range(10)]
print(squares)

# Normal way
squares = []
for n in range(10):
    squares.append(n**2)
print(squares)

# It is possible to add a condition 'if'
short_companies = [company for company in companies if len(company) < 6]
print(short_companies)
```

## 6. External libraries: NumPy and Matplotlib

External libraries are useful tools. The keyword `import` imports a library, or some of its modules. It is very common to give the library a shorter name with the keyword `as`.

We look at three libraries for now:

- math: mathematical functions
- numpy: support for large, multi-dimensional arrays and matrices
- matplotlib: visualization

**Example:**

```python
# Import the library
import math

print(math.cos(5))

# Import modules from a library
from math import sin, cos

print(cos(5), sin(4))
```

```python
# Used to embed plots inside an ipython notebook
%matplotlib inline
# Import the pyplot module from matplotlib library, then rename it as plt
import matplotlib.pyplot as plt

# Really simple example:
y = [1, 2, 3, 4, 5, 4, 3, 2, 1]
x = [1, 2, 3, 4, 5, 6, 7, 8, 9]
plt.title("Our graph test")
plt.plot(x, y)
plt.plot([1, 2, 3, 4])
plt.ylabel("some numbers")
plt.show()
```

```python
import numpy as np
import math as math

X = np.linspace(0, 10, 100001)  # Create 100001 values from 0 to 10
Y = []

for x in X:
    y = math.sin(x)
    Y.append(y)

plt.plot(X, Y, "mx")
plt.title("The Sine Wave")
plt.xlabel("X")
plt.ylabel("sin(X)")
```

We can also draw bar plots and histograms:

```python
vals = [7, 6.2, 3, 5, 9]
xval = [1, 2, 3, 4, 5]
plt.bar(xval, vals)
plt.show()

Y = []
for x in range(0, 100000):
    Y.append(np.random.randn())

plt.hist(Y, 100)
plt.show()
```
