## Introduction

Notebooks represent an interactive way of sharing knowledge and code. Here we review some elementary ways of creating your notebooks. We will be using Google's Colab, which does not require any installations and it runs on a browser.

A notebook consists of cells that contain either **code** or **markdown (text)**. Markdown is very similar to HTML (which serves to write text in a web page), but simplified. Here are two places to learn it:
- Try markdown on your own in [Google Colab](https://colab.research.google.com/notebooks/markdown_guide.ipynb)
- You can also find a more detailed and [lengthy introduction](https://handbook.gitlab.com/docs/markdown-guide/).

### Code vs Text/Markdown

By default, a cell contains code.
- If you want the cell to understand markdown, you can convert it to a text cell by pressing `control M M` on a Mac (that is hold down the `control` key and press two times the letter `M`).
- You can convert a cell into a code cell using the shortcut `control M Y`.

## Basic Markdown

Assuming your cell is in markdown mode, to create a title you use the `#` (hash) symbol

```# My Project for this class```

will give

# My Project for this class

To create a smaller heading you use more hashes

```### Introduction to the project```

### Introduction to the project

It is also very easy to create numbered or unnumbered lists as follows

```
- First item
- Second item
- ...
```

giving
- First item
- Second item
- ...

or replace the - with numbers 1. 2. 3. for a numbered list

If you want to create a second level indentation you can put 4 spaces before the `-` symbol, such as:

- Bullet level 1
    - Bullet level 2

## Interactive Components

You can connect your existing Python code with interactive components, such as buttons, sliders, ...

The following code will convert inches to cm, using an interactive slider.

**Interactive elements may not work in every environment. If the slider does not appear, you can skip this part.**

````python
from ipywidgets import interact


def inches_to_cm(inches=2):
    return "%.2f inches is %.2f cm" % (inches, inches * 2.54)


interact(inches_to_cm, inches=(0, 8, 1));
````

You can find a list of available widgets [here](https://ipywidgets.readthedocs.io/en/stable/examples/Widget%20List.html) and you may read an article with more [advanced applications](https://towardsdatascience.com/interactive-controls-for-jupyter-notebooks-f5c94829aee6) of the widgets.


## Graphs

In a notebook you can also embed graphs from the analysis of your dataset, including barcharts, scatterplots, piecharts, jointplots and others.

````python
%matplotlib inline
import matplotlib.pyplot as plt

plt.rcdefaults()
import numpy as np
import matplotlib.pyplot as plt

objects = ("Python", "C++", "Java", "Perl", "Scala", "Lisp")
y_pos = np.arange(len(objects))
usage = [10, 8, 6, 4, 2, 1]

plt.bar(y_pos, usage, align="center", alpha=0.5)
plt.xticks(y_pos, objects)
plt.ylabel("Usage")
plt.title("Programming language usage")

plt.show()
````

Many times, you will try various versions of your Python code and try to optimize it, i.e., make it run faster. How do you know that some different code is faster than your existing one?

A very useful 'magic' command is the ```timeit``` which lets you measure the time it takes to run a command, or all the commands in a cell. Such a command will help you try different versions of your code and identify the most efficient variant. You can reach the same destination via multiple paths, but some are faster than others!
