## 1. The Arithmetic Mean

$$ Mean(x) = \frac{1}{n} \sum_{i=1}^{n} x_i$$

An embedded, formatted code block:
    
```python

```

````python
# actual python calculation
x = [1, 3, 9, 4]
x_bar = 1 / len(x) * sum(x)
x_bar
````
