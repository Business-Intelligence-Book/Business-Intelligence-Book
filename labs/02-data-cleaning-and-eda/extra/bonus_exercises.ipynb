* Fix the data types. In particular, turn Yes/No values into the categories 1 and 0. The first step is to find the unique values of each column; then we convert them to numbers.

Let's first look at the `tenure` column. The tenure is the number of months the customer has stayed with the company, so it is important to check for null or 0 values.

**To do:** access the column with `customers.tenure` or `customers['tenure']`, and count the rows where `tenure == 0`. To select the rows that satisfy a condition, use `.loc[]`: `customers.loc[customers.tenure == 0]`.

**Note:** these 11 rows will be deleted later. They are probably customers who signed a contract and cancelled it before it started, or the data is simply missing. They will be dropped in the next steps (check for empty strings, NaN values, etc.).

The following function identifies the unique values of each column. Its 2 arguments give the range of the number of unique values you are looking for. It returns the columns that have this number of unique values (this will be useful later to clean the columns).

```python
def identify_uniques(min_, max_):
    columns = list(customers.columns)
    print(columns, "\n")
    interested_columns = []
    for col in columns:
        # Get the unique values of the column
        uniques = customers[col].unique().tolist()
        # Check whether their number is in the range
        if len(uniques) >= min_ and len(uniques) < max_:
            print(col, " uniques values: ", uniques)
            interested_columns.append(col)
    return interested_columns
```

Let's look at the binary and multi-class columns.

```python
print("Binary classes: ")
interested_columns = identify_uniques(0, 4)

print("Multi classes: ")
identify_uniques(3, 15)
```

**Note:** several columns contain only two categories. Some of them also contain the values "No internet service" and "No phone service", which can be converted directly to "No". There are also columns with a few more classes: *InternetService*, *Contract* and *PaymentMethod*.

## 3. Statistical analysis

To choose the features of a future model, we look at statistical indicators such as the correlation between variables, especially with the *Churn* column, which will be the target variable.

* Correlation between the columns

The Pearson correlation coefficient is the most widely used. It measures the strength of the linear relationship between two variables. When the relationship is not linear, or the variables are far from normally distributed, the Spearman rank correlation can be more appropriate.

**To do:**

**Note:** we are interested in the columns that are strongly correlated with churn, and in those that are not. The main features related to churn are *Contract*, *tenure*, *PaymentMethod*, *TotalCharges*, *SeniorCitizen*, *MonthlyCharges* and *PaperlessBilling*. Note that the order of the two most correlated features depends on the method.

**Note:**
These results are consistent with some hypotheses:

**To do:** run the following code. It plots the correlation matrix with `sns.heatmap()`.

```python
corr = customers.corr()

plt.figure(figsize=(10, 10))
plt.title("Correlation matrix")
ax = sns.heatmap(
    corr, vmin=-1, vmax=1, center=0, cmap=sns.cubehelix_palette(200), square=True
)
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, horizontalalignment="right")
```
